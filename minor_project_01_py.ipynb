```python
%%writefile pailib.py

def cal_tpi(coding_min, valid_days):
  return sum(coding_min) / valid_days

def cal_aai(study_min, class_min, valid_days):
  return (sum(study_min) + sum(class_min)) / valid_days

def cal_phai(fitness_min, valid_days):
  return sum(fitness_min) / valid_days

def cal_sri(sleep_min, valid_days):
  return sum(sleep_min) / valid_days

def cal_abi(free_uncounted_min, valid_days):
  return sum(free_uncounted_min) / valid_days

def cal_tui(tracked_min, valid_days):
  return sum(tracked_min) / valid_days

def cal_ei(days_feeling, satisfaction_level, energy_level, valid_days):
  feelings = {
    "Excellent": 5,
    "Good": 4,
    "Neutral": 3,
    "Low": 2,
    "Stressed": 1
  }

  satisfaction = {
    "Very Satisfied": 5,
    "Satisfied": 4,
    "Neutral": 3,
    "Unsatisfied": 2,
    "Very Unsatisfied": 1
  }

  energy = {
    "High": 3,
    "Medium": 2,
    "Low": 1
  }

  days_feeling = list(map(lambda x: feelings[x], days_feeling))
  satisfaction_level = list(map(lambda x: satisfaction[x], satisfaction_level))
  energy_level = list(map(lambda x: energy[x], energy_level))

  return (sum(days_feeling) + sum(satisfaction_level) + sum(energy_level)) / (3 * valid_days)

def cal_dci(recorded_days, excepted_days):
  return recorded_days * 100 / excepted_days

def cal_pai(tpi, aai, phai, sri, tui, ei, dci):
  return (0.15 * tpi) + (0.20 * aai) + (0.15 * phai) + (0.20 * sri) + (0.15 * tui) + (0.10 * ei) + (0.05 * dci)
```

```python
import pailib
import openpyxl

wb = openpyxl.load_workbook("12600512.xlsx", data_only=True)
ws = wb["Daily Log"]

recorded_days = ws.max_row
valid_days = 40

sleep_min = []
fitness_min = []
study_min = []
coding_min = []
class_min = []
classes_attended = []
total_tracked_min = []
free_unaccounted_min = []
days_feeling = []
satisfaction_level = []
energy_level = []


for column in range(2, 14):
  for row in range(6, ws.max_row + 1):
    value = ws.cell(row, column).value
    if value is not None:
      if column == 2:
          sleep_min.append(value)
      elif column == 3:
          fitness_min.append(value)
      elif column == 4:
          study_min.append(value)
      elif column == 5:
          coding_min.append(value)
      elif column == 6:
          class_min.append(value)
      elif column == 7:
          classes_attended.append(value)
      elif column == 9:
          total_tracked_min.append(value)
      elif column == 10:
          free_unaccounted_min.append(value)
      elif column == 11:
          days_feeling.append(value)
      elif column == 12:
          satisfaction_level.append(value)
      elif column == 13:
          energy_level.append(value)


tpi = pailib.cal_tpi(coding_min, valid_days)
aai = pailib.cal_aai(study_min, class_min, valid_days)
phai = pailib.cal_phai(fitness_min, valid_days)
sri = pailib.cal_sri(sleep_min, valid_days)
tui = pailib.cal_tui(total_tracked_min, valid_days)
ei = pailib.cal_ei(days_feeling, satisfaction_level, energy_level, valid_days)
dci = pailib.cal_dci(recorded_days, valid_days)

abi = pailib.cal_abi(free_unaccounted_min, valid_days)

pai = pailib.cal_pai(tpi, aai, phai, sri, tui, ei, dci)

print(f"Tech Productivity Index is {tpi}")
print(f"Academic Activity Index is {aai}")
print(f"Physical Activity Index is {phai}")
print(f"Sleep & Recovery Index is {sri}")
print(f"Activity Balance Index is {abi}")
print(f"Time Utilization Index is {tui}")
print(f"Experience Index is {ei:.2f}")
print(f"Data Continuity Index is {dci:.2f}")
print(f"Personal Activity Index is {pai:.2f}")
```
