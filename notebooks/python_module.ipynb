### Create a Python module named `spending` to track your daily expenses.


The module should contain the following functions:
```python
morning_spend()
```
Record or calculate the amount spent during the morning.
```python
evening_spend()
```
Record or calculate the amount spent during the evening.
```python
night_spend()
```
Record or calculate the amount spent during the night.
```python
total_spending()
```
Calculate and return the total spending for the day by adding the morning, evening, and night expenses.

````python
%%writefile spendings.py

morning_spending = 0
evening_spending = 0
night_spending = 0
total_spending = 0

def input_morning_spend(spending = 10):
  global morning_spending
  morning_spending = spending

def input_evening_spend(spending):
  global evening_spending
  evening_spending = spending

def input_night_spend(spending):
  global night_spending
  night_spending = spending

def get_total_spend():
  global total_spending
  total_spending = morning_spending + evening_spending + night_spending

  return total_spending

````

````python
import spendings


if __name__ == "__main__":

  morning_spending = int(input("Enter your morning spend: "))
  spendings.input_morning_spend(morning_spending)

  evening_spending = int(input("Enter your evening spend: "))
  spendings.input_evening_spend(evening_spending)

  night_spending = int(input("Enter your night spend: "))
  spendings.input_night_spend(night_spending)

  total_spend = spendings.get_total_spend()
  print(f"\nToday you have spend {total_spend} rupees")

````
