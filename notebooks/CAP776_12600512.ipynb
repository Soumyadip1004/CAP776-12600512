##Sandbox2
Let's consider a scenario where you want to distribute prizes evenly among a group of students, and you need to find out how many prizes will be left over after distribution. Consider you have 53 prizes and 8 students. You want to distribute the prizes equally among the students and find out how many prizes will be left after each student gets their share.

```python
number_of_prizes = 53
number_of_students = 8

distributed_prizes = number_of_prizes // number_of_students
left_over_prizes = number_of_prizes % number_of_students

print(f"Number of Prizes distributed to each student: {distributed_prizes}")
print(f"Remaining prizes: {left_over_prizes}")
```
